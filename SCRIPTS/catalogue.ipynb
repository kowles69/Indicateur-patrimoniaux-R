```python
import PyPDF2
import io
import os
from datetime import datetime
```

```python
def valid_files(pdf_files):
    ''' prend une liste de fichiers PDF, vérifie leur existence sur le disque et retourne une liste de fichiers valides. 
    Si aucun fichier n'est valide, l'opération est annulée.'''

    valid_files = []
    for pdf in pdf_files:
        if os.path.exists(pdf):
            valid_files.append(pdf)
        else:
            print(f"Le fichier {pdf} n'existe pas et sera ignoré.")

    if not valid_files:
        print("Aucun fichier valide à concaténer. Opération annulée.")
        return
    else:
        print('Validation terminée')
        return(valid_files)

def concat_pdf(output_path, valid_files):
    '''usionne les fichiers PDF valides en un seul fichier PDF de sortie. 
    Chaque fichier est ouvert en mode binaire, ajouté à un objet PdfMerger, puis le résultat est écrit dans un fichier de sortie.'''

    merger = PyPDF2.PdfMerger()
    for pdf in valid_files:
        with open(pdf, 'rb') as f:
            merger.append(f)

    with open(output_path, 'wb') as output_file:
        merger.write(output_file)

if __name__ == "__main__":
    # Liste des fichiers PDF à concaténer 
    racine = 'L:/EGDD/DAEA/PAG/K_STAGES/2023_Amen Allah NOUIRA/Indicateurs/catalogue/'
    path_input = f'{racine}0_INPUT/'
    path_fiches = f'{path_input}FICHES/'
    pdf_files = [
        f'{path_input}page_de_garde.pdf',
        f'{path_input}Intro_Sommaire.pdf',
        f'{path_input}metropole.pdf',
        f'{path_fiches}Fiche_patrimoine_EU_U_Metropole_.pdf',
        f'{path_fiches}Fiche_patrimoine_EP_Metropole_.pdf'
    ]
    dic_niveau_extraction={
        "Commune": ['ALBIGNY-SUR-SAONE', 'BRON', 'CAILLOUX-SUR-FONTAINES', 'CALUIRE ET CUIRE', 'CHAMPAGNE-AU-MONT-D OR', 'CHARBONNIERES-LES-BAINS', 'CHARLY', 'CHASSIEU', 'COLLONGES-AU-MONT- D OR', 'CORBAS', 'COUZON-AU-MONT-D OR', 'CRAPONNE', 'CURIS-AU-MONT-D OR', 'DARDILLY', 'DECINES CHARPIEU', 'ECULLY', 'FEYZIN', 'FLEURIEU-SUR-SAONE', 'FONTAINES SUR SAONE', 'FONTAINES-SAINT-MARTIN', 'FRANCHEVILLE', 'GENAY', 'GIVORS', 'GRIGNY', 'IRIGNY', 'JONAGE', 'LA MULATIERE', 'LA TOUR DE SALVAGNY', 'LIMONEST', 'LISSIEU', 'LYON 1ER', 'LYON 2EME', 'LYON 3EME', 'LYON 4EME', 'LYON 5EME', 'LYON 6EME', 'LYON 7EME', 'LYON 8EME', 'LYON 9EME', 'MARCY-L ETOILE', 'MEYZIEU', 'MIONS', 'MONTANAY', 'NEUVILLE SUR SAONE', 'OULLINS', 'PIERRE BENITE', 'POLEYMIEUX-AU-MONT-D OR', 'QUINCIEUX', 'RILLIEUX LA PAPE', 'ROCHETAILLEE-SUR-SAONE', 'SAINT CYR AU MONT D OR', 'SAINT-DIDIER-AU-MONT-D OR', 'SAINT-GENIS-LAVAL', 'SAINT-GENIS-LES-OLLIERES', 'SAINT-GERMAIN-AU-MONT-D OR', 'SAINT-ROMAIN-AU-MONT-D OR', 'SAINTE FOY LES LYON', 'SATHONAY-CAMP', 'SATHONAY-VILLAGE', 'SOLAIZE', 'ST FONS', 'ST PRIEST', 'TASSIN-LA-DEMI-LUNE', 'VAULX EN VELIN', 'VENISSIEUX', 'VERNAISON', 'VILLEURBANNE'],
        "Systeme Assainissement": ['Genay', 'Fontaines', 'Jonage', 'La Feyssine', 'Lissieu Le Bourg', 'Lissieu Sémanet', 'Meyzieu', 'Neuville', 'Pierre-Bénite', 'Quincieux', 'Saint-Fons', 'Saint-Germain'],
        "Bassin versant": ['GENAY, GRAND RIEUX', 'MAINTENU, COMBE, GRANDES GORGES', 'THOU', 'TORRIERES', 'ECHETS', 'VOSGES', 'RAVIN', 'PINAY DE ST ROMAIN, MONGELAS, CHARBOTTES', 'ROCHECARDON', 'SEMANET, GORGES, MALIGNEUX, AZERGUES', 'PLANCHES', 'MOUCHE', 'VERNIERES, FEE DES EAUX, SERVE, RAZAT', 'YZERON', 'GUA, EPI, RIZAN, BLETTA', 'GARON', 'CHANAUX', 'OZON', 'GIER', 'MERDARY, COMBE DE BANS', 'SAONE', 'RHONE', 'RIZE', 'RATAPON'],
        "Captage": ['Captage de Chassieu', 'Captage de Corbas Romanettes', 'Captage de Crépieux-Charmy', 'Captage de Curis', 'Captage de Dècines', 'Captage de Fleurieu', 'Captage de Jonage', 'Captage de Meyzieu', 'Captage de Mions', 'Captage de St Priest', 'Lac de Miribel-Jonage']
    }
    dic_sep = {
        "Commune": f'{path_input}COMMUNES.pdf',
        "Systeme Assainissement": f'{path_input}SysAss.pdf',
        "Bassin versant": f'{path_input}BV.pdf',
        "Captage": f'{path_input}Captage.pdf'
    }
    dic_cartes = {
        "Commune": f'{path_input}carte_communes.pdf',
        "Systeme Assainissement": f'{path_input}carte_SA.pdf',
        "Bassin versant": f'{path_input}carte_BV.pdf',
        "Captage": f'{path_input}Carte_Captage.pdf'
    }
    for zone in dic_niveau_extraction:
        pdf_files.append(dic_sep[zone])
        pdf_files.append(dic_cartes[zone])
        for element in dic_niveau_extraction[zone]:
            fiche_EU_file = f'{path_fiches}Fiche_patrimoine_EU_U_{zone}_{element}.pdf'
            fiche_EP_file = f'{path_fiches}Fiche_patrimoine_EP_{zone}_{element}.pdf'
            pdf_files.append(fiche_EU_file)
            pdf_files.append(fiche_EP_file)
    valid_files = valid_files(pdf_files)
    # Nom du fichier PDF de sortie
    output_path = f'{racine}3_OUTPUT/catalogue_Patrimoine_DCE_{datetime.now().month}_{datetime.now().year}.pdf'

    concat_pdf(output_path, valid_files)

    print(f"Les fichiers PDF ont été concaténés avec succès dans {output_path}")
```
